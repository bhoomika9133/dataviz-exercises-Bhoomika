## Scatter & Bubble Charts: Gapminder

> **Push to:** `week04/lecture04_exercise.ipynb`

**Rules:**
1. Colour used **sparingly** — one categorical variable, no rainbow
2. If showing all continents, either use accessible palette OR grey all + highlight one
3. `size_max` set when using bubble size
4. Log scale for GDP per capita
5. Insight title

---


```python
import pandas as pd
import plotly.express as px


# Dataset: Gapminder — GDP, Life Expectancy, Population by Country
# Source: Gapminder Foundation (gapminder.org)

df = px.data.gapminder()
print(f"Loaded: {len(df)} rows")
print(df.head())
```

```python
# explore

print(df.info())
print("Years:", sorted(df['year'].unique()))
print("Continents:", df['continent'].unique())
print(df.describe().round(1))
```

## Task 1 — Scatter: life expectancy change over time

**What to build:** A scatter showing **GDP per capita vs life expectancy** for **two years** (2002 and 2007) to show how both moved — use **colour for year** (just 2 colours), **one continent only**.

Choose any continent except Africa (that was the example). Highlight the change direction.

> 💡 Filter: `df.loc[df['continent'] == 'YOUR_CHOICE']` then filter years


```python
import plotly.express as px

americas = df.loc[df['continent'] == 'Americas']
americas = americas.loc[americas['year'].isin([2002, 2007])]

fig = px.scatter(
    americas,
    x='gdpPercap',
    y='lifeExp',
    color='year',
    hover_name='country',
    log_x=True,
    color_discrete_map={
        2002: 'steelblue',
        2007: 'tomato'
    },
    labels={
        'gdpPercap': 'GDP per Capita (log scale)',
        'lifeExp': 'Life Expectancy (years)',
        'year': 'Year'
    },
    title='Americas: GDP per Capita and Life Expectancy, 2002–2007',
    template='plotly_white'
)

fig.update_traces(marker=dict(size=10, opacity=0.8))
fig.update_layout(height=550)

fig.show()
```

## Task 2 — Bubble chart: tell a story

**What to build:** A bubble chart (full 2007 dataset, all countries) where:
- x = GDP per capita (log scale)
- y = life expectancy
- size = population
- colour = ONE continent highlighted (your choice), all others grey
- At least one annotation explaining the highlighted group's story

> This is the grey-and-highlight technique applied to a bubble chart.


```python
import plotly.express as px

df_2007 = df[df["year"] == 2007].copy()

df_2007["group"] = df_2007["continent"].apply(
    lambda continent: "Europe" if continent == "Europe"
    else "Other continents"
)

fig = px.scatter(
    df_2007,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    color="group",
    hover_name="country",
    hover_data={
        "continent": True,
        "gdpPercap": ":,.0f",
        "lifeExp": ":.1f",
        "pop": ":,.0f"
    },
    color_discrete_map={
        "Europe": "#377EB8",
        "Other continents": "#BDBDBD"
    },
    log_x=True,
    size_max=40,
    opacity=0.8,
    labels={
        "gdpPercap": "GDP per Capita (USD, log scale)",
        "lifeExp": "Life Expectancy (years)",
        "pop": "Population",
        "group": "Continent"
    },
    title="Europe in 2007: Wealth and Life Expectancy",
    template="plotly_white"
)

fig.add_annotation(
    x=20000,
    y=75,
    text="European countries generally show<br>higher life expectancy and income",
    showarrow=True,
    arrowhead=2,
    ax=70,
    ay=-50,
    bgcolor="white",
    bordercolor="#377EB8",
    borderwidth=1
)

fig.update_layout(
    height=600,
    width=1000,
    legend_title_text="",
    margin=dict(l=70, r=40, t=90, b=70)
)

fig.show()
```
